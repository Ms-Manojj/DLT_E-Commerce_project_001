
# 🚀 End-to-End E-Commerce Order Data Engineering Pipeline

## 1. Project Overview

The objective of this project is to build an end-to-end E-Commerce Order Data Engineering Pipeline using Azure Databricks.

This project will simulate a real-world e-commerce order processing system, starting from raw data ingestion and progressing through incremental loading, data quality validation, historical data management, data transformation, and finally delivering business-ready data.

All source datasets will be stored and accessed directly through **Databricks Volumes**.

## 2. Source Datasets

We will primarily work with the following three datasets:

| Dataset | Description |
|---|---|
| Customer | Contains customer information and profile details. |
| Orders | Contains order transactions, customer references, order dates, and order amounts. |
| Status | Contains order status information and status history. |

## 3. Key Project Implementations

### 3.1 Incremental Data Loading
- Implement incremental data ingestion.
- Process only new and updated records.
- Use Databricks Auto Loader.
- Maintain checkpoints to avoid duplicate processing.

### 3.2 Slowly Changing Dimension (SCD Type 2)
- Implement SCD Type 2 to maintain historical customer changes.
- Track changes using effective start date and end date.
- Maintain a current record indicator.
- Preserve historical records for auditing and reporting.

### 3.3 Data Quality Framework
- Implement data validation rules.
- Handle NULL values and duplicate records.
- Validate primary keys and referential integrity.
- Identify and isolate invalid records.
- Implement appropriate error-handling mechanisms.

### 3.4 Data Transformation & Integration
- Clean and standardize source datasets.
- Integrate Customer, Orders, and Status datasets.
- Apply business rules and transformation logic.
- Create reusable transformation processes.

### 3.5 Gold Layer – E-Commerce Order Data Mart
- Create a centralized Order Data Mart.
- Integrate customer, order, and status information.
- Maintain complete order-level details.
- Prepare business-ready datasets for reporting and analytics.

## 4. Project Architecture

We will follow the **Medallion Architecture**:

| Layer | Description |
|---|---|
| Bronze | Raw data ingestion from Databricks Volumes. |
| Silver | Data cleansing, validation, deduplication, SCD Type 2, and transformations. |
| Gold | Integrated and business-ready E-Commerce Order Data Mart. |

## 5. Technology Stack

- Azure Databricks
- PySpark
- Spark SQL
- Delta Lake
- Unity Catalog
- Databricks Volumes
- Databricks Auto Loader

## 6. Expected Outcome

At the end of this project, we will have a complete, scalable, and incremental E-Commerce Data Engineering Pipeline that converts raw source files into a business-ready Order Data Mart.

The final Gold layer will contain all relevant customer and order information along with the latest order status and historical status details, enabling business reporting and analytical use cases.
